# 🤖 Personal RAG Assistant Demo

This notebook demonstrates the full RAG pipeline using the modular source code in the `src/` directory.

```python
import sys
from pathlib import Path

# Add src directory to path
sys.path.append(str(Path("../src")))

from processor import DocumentProcessor
from vector_store import VectorStore
from rag_engine import RAGEngine

print("Modules imported successfully!")
```

## Step 1: Document Indexing
In this step, we extract text from a PDF, chunk it, and store it in ChromaDB.

```python
# Initialize Processor and Vector Store
processor = DocumentProcessor(chunk_size=500, overlap=100)
store = VectorStore(db_path="./chroma_db")

# Path to your PDF
pdf_path = "../data/sample_docs/resume.pdf" 

# Process PDF
text = processor.extract_text_from_pdf(pdf_path)
chunks = processor.chunk_text(text)

# Store in Vector DB
store.add_documents(chunks, metadata_source="My Resume")

print(f"Indexed {len(chunks)} chunks from {pdf_path}")
```

## Step 2: RAG Chatting
Now we use the `RAGEngine` to ask questions grounded in the indexed documents.

```python
# Initialize RAG Engine
rag = RAGEngine(vector_store=store)

chat_history = []
query = "What are the key technical skills of the person?"

answer = rag.generate_answer(query, chat_history)
print(f"Question: {query}\nAnswer: {answer}")
```
